Question 3: You wrote some important notes in a file called notes.txt. Your senior developer
asks you to check what is written in that file quickly, without opening any editor. What command
will you use and why?
- To check the contents of a file called notes.txt quickly without opening any text editor, I would use the command cat notes.txt. The cat command stands for “concatenate” and is commonly used in Linux and Unix-based systems to display the contents of a file directly in the terminal. When the command cat notes.txt is executed, the system reads the file and immediately prints all the text written inside it on the screen. This method is very useful for developers because it is fast, simple, and saves time while working in the command line environment. Instead of opening applications like VS Code or Notepad, the user can quickly inspect the file directly from the terminal. It is especially helpful for small text files such as notes, logs, or configuration files. For example, if the file contains project instructions or important reminders, the cat notes.txt command will instantly show them in the terminal window. Therefore, this command is widely used by developers for quick file viewing and efficient workflow management.

Question 4: You are setting up a web server on a Linux machine. You run your server script but
it says 'Permission denied'. Your senior tells you the file needs to be executable. What will you
do to fix this?
- If I run a web server script on a Linux machine and the terminal shows a “Permission denied” error, it usually means that the script file does not have executable permission. In Linux, every file has permissions that control who can read, write, or execute it. Since the system is not allowing the script to run, I need to make the file executable first.

- To fix this problem, I would use the command chmod +x filename. For example, if my server script file is named server.sh, I would type chmod +x server.sh in the terminal. The chmod command stands for “change mode,” and the +x part gives execute permission to the file. After running this command, the system allows the script to be executed like a program.

- Once the permission is given, I can run the script using ./server.sh. Here, ./ tells Linux to execute the file from the current directory. Before adding executable permission, the operating system blocks the file for security reasons, which is why the “Permission denied” message appears.

- This process is very important while working with shell scripts, web servers, or deployment files in Linux environments. Developers often face this issue when downloading scripts from another system or creating new files because new files may not automatically have execute permission. By using the chmod +x command, the problem is solved quickly and the server script can run properly.

Question 5: You have a large log file called app.log and you need to find all lines that contain
the word 'error'. The file has thousands of lines and you cannot read it manually. What
command will you use?
- If I have a large log file called app.log and I need to find all the lines that contain the word “error,” I would use the command grep error app.log. The grep command is used in Linux to search for specific words or patterns inside a file. It is very useful when a file contains thousands of lines and reading everything manually would take too much time.

- When I type grep error app.log, the system searches through the entire app.log file and displays only the lines where the word “error” appears. This makes it easy to quickly find problems in the log file without checking every line one by one.

- This command is very important for developers and system administrators because log files often store information about system activities, warnings, and errors. When an application crashes or does not work properly, checking the log file helps identify the issue. Since log files can be very large, using grep saves time and makes troubleshooting faster.

- For example, if the log file contains server messages and some of them show connection failures or application errors, grep error app.log will show only those related lines. This helps the developer focus directly on the problem and fix it quickly.

- So, the grep error app.log command is the best choice because it is fast, efficient, and designed for searching specific information inside large files. It is a common command used in real-world Linux development and server management.

Question 6: You are running a command that generates a long list of files. You want to see only
the files whose names contain the word 'config'. How will you combine two commands to do this
in one line?
- If I run a command that generates a very long list of files and I only want to see the files whose names contain the word “config,” I can combine two commands using a pipe operator (|). The command I would use is ls | grep config.

- In this command, ls is used to display the list of files and folders in the current directory. Since the output may contain many file names, I use the pipe operator (|) to send that output directly to another command. The second command, grep config, searches through the list and shows only the lines that contain the word “config.”

- The pipe operator is very useful in Linux because it allows one command output to become the input for another command. Instead of checking hundreds of file names manually, the system automatically filters the results and displays only the required files.

- For example, if the directory contains files like config.txt, appconfig.json, data.txt, and server.log, the command ls | grep config will display only config.txt and appconfig.json because those names contain the word “config.”

- This method is fast, efficient, and commonly used by developers and system administrators while working with large numbers of files. It saves time, improves productivity, and makes searching for specific files much easier in the command line environment.

Question 7: A company's website is not working. The developer checks the HTML file and sees
there is no DOCTYPE declaration at the top and the content is placed directly without any head
or body tags. Why could this be a problem?
- If a company’s website does not have a DOCTYPE declaration at the top and the content is written directly without proper head and body tags, it can create many problems in how the webpage works and appears in browsers. The DOCTYPE declaration is important because it tells the browser which version of HTML is being used. Without it, the browser may enter “quirks mode,” where it tries to display the page using old and inconsistent rules. This can cause the website layout, fonts, spacing, and design to look different in different browsers.

- The head and body tags are also essential parts of a proper HTML structure. The head section contains important information about the webpage, such as the page title, meta tags, CSS links, and scripts. Without the head tag, the browser may not load stylesheets, fonts, or important metadata correctly. This can make the website look broken or reduce performance and SEO.

- The body tag contains all the visible content of the webpage, such as text, images, buttons, and forms. If the content is placed directly without a body tag, the browser may still try to display it, but the structure becomes invalid and unpredictable. Different browsers may handle the page differently, which can lead to display issues or functionality problems.

- Using proper HTML structure is very important in web development because it improves browser compatibility, readability, maintenance, accessibility, and search engine optimization. A correctly structured HTML document usually starts with <!DOCTYPE html>, followed by html, head, and body tags. Without these elements, the website may not behave correctly and can create a poor user experience

Question 9: What is the difference between > and >> in the terminal? Give one example of
when you would use each.
- In the terminal, both > and >> are used for output redirection, which means sending the output of a command into a file instead of showing it on the screen. However, both symbols work differently.

- The > operator is used to overwrite the contents of a file. If the file already contains some data, using > will remove the old content and replace it with the new output. If the file does not exist, the terminal will create it automatically. For example, if I use the command echo Hello > notes.txt, the word “Hello” will be written into the notes.txt file. If the file already had previous text inside it, that text will be deleted and replaced with “Hello.” This operator is useful when I want to create a new file or completely replace old content with updated information.

- On the other hand, the >> operator is used to append data to a file. It adds the new output at the end of the existing content without deleting anything already present in the file. For example, if I use echo Hello >> notes.txt, the word “Hello” will be added at the bottom of the file. If the file already contains notes or logs, they will remain unchanged and the new text will simply be added after them. This is very useful for log files, reports, or situations where I want to keep old data and continuously add new information.

- In simple words, > replaces file content, while >> adds content to the existing file. Developers and system administrators use these operators frequently while working in Linux because they help manage files efficiently through the command line.

Question 10: What is the difference between rm and rm -r? Why should you be careful when
using rm -r?
- The rm command in Linux is used to delete files from the system. It permanently removes files from a directory through the terminal. For example, if I use the command rm notes.txt, the file named notes.txt will be deleted from the current directory.

- On the other hand, rm -r is used to delete directories and everything inside them recursively. The -r option stands for “recursive.” This means the command will go inside the folder, remove all files and subfolders, and then delete the main folder itself. For example, if I use rm -r project, the entire project folder along with all its contents will be permanently deleted.

- The main difference between rm and rm -r is that rm normally removes only files, while rm -r removes complete folders and their contents.

- We should be very careful while using rm -r because it can permanently erase a large amount of important data very quickly. Unlike the recycle bin in Windows, deleted files in Linux usually cannot be recovered easily after using rm -r. If the wrong folder name is typed by mistake, important project files, system files, or documents may be lost permanently.

- For example, accidentally running a command like rm -r important-folder can completely remove all project data inside that folder. That is why developers often double-check the folder name before executing the command.

- Therefore, rm -r is a powerful command and should be used carefully because it deletes directories recursively and permanently without easy recovery.

Question 11: What is the difference between semantic HTML tags like section and article
versus a generic div tag? Why does it matter?
- Semantic HTML tags like section and article are used to describe the meaning and purpose of the content inside them, while a div tag is a generic container that does not give any meaning about the content. The main difference is that semantic tags tell both the browser and developers what kind of content is present, whereas a div is mainly used for grouping elements for styling or layout purposes.

- For example, the section tag is used to group related content together, such as a services section or product section on a webpage. Thearticle tag is used for independent content like a blog post, news article, or product card that can stand on its own. When developers use these semantic tags, the structure of the webpage becomes more meaningful and organized.

- On the other hand, a div tag does not explain anything about the content inside it. It simply acts as a box or container. Developers often use div for CSS styling, layouts, or JavaScript functionality when no specific semantic tag is suitable.

- Using semantic HTML matters for many reasons. First, it improves readability and maintainability of the code because developers can quickly understand the webpage structure. Second, it helps search engines understand the content better, which improves SEO (Search Engine Optimization). Third, semantic tags improve accessibility because screen readers and assistive technologies can identify different parts of the webpage more accurately for users with disabilities.

- For example, if everything on a webpage is written only inside div tags, the structure becomes confusing and less meaningful. But using tags like header, section, article, and footer makes the webpage clearer and more professional.

- Therefore, semantic HTML is important because it gives proper meaning to webpage content, improves accessibility and SEO, and makes the code easier to manage compared to using only generic div tags.

Question 12: Why are forms important in web development? Name three input types used in
HTML forms and explain what each is used for.
- Forms are very important in web development because they allow users to enter and send data to a website or application. Websites use forms to collect information such as login details, contact messages, feedback, registrations, payments, and search queries. Without forms, users would not be able to interact properly with websites. Forms act as a communication method between the user and the server, helping websites receive and process user information.

- In HTML, forms are created using the form tag, and different types of input fields are used depending on the kind of data needed from the user.

- One commonly used input type is text. It is created using input type="text" and is used for entering short text information such as a user’s name, city, or username. It allows users to type normal text into a single-line input box.

- Another important input type is email. It is created using input type="email" and is specifically used for collecting email addresses. This input type helps with validation because the browser checks whether the entered value follows a proper email format, such as including @gmail.com.

- The third common input type is password. It is created using input type="password" and is used for entering secure information like passwords or PINs. The characters typed in this field are hidden or displayed as dots or stars to protect user privacy.

- These input types make forms more useful, user-friendly, and secure. Different input fields help websites collect the correct type of information efficiently. Therefore, forms play a major role in creating interactive and functional websites in web development.
